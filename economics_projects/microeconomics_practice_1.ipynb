```python
#код для пункта b)
def analyze_consumer_properties():
    consumer_properties = {}
    for customer in df.columns[1:]:
        customer_data = df[customer].values
        max_willingness = customer_data[0]
        total_consumption = len([p for p in customer_data if p > 0])
        max_total_value = sum(customer_data[:total_consumption])
        if total_consumption > 1:
            elasticity_measure = (customer_data[total_consumption-1] - customer_data[0]) / (total_consumption - 1)
        else:
            elasticity_measure = 0
        units_at_p20 = len([p for p in customer_data if p >= 20])
        units_at_p10 = len([p for p in customer_data if p >= 10])
        if customer_data[0] > 0:
            steepness = (customer_data[0] - customer_data[min(4, total_consumption-1)]) / customer_data[0]
        else:
            steepness = 0
        consumer_properties[customer] = {
            'max_willingness': max_willingness,
            'max_total_value': max_total_value,
            'elasticity': elasticity_measure,
            'units_p20': units_at_p20,
            'units_p10': units_at_p10,
            'steepness': steepness,
            'total_units': total_consumption
        }
    return consumer_properties
def create_proper_groups(consumer_props):
    group_high = []
    group_medium = []
    group_low = []
    for customer, props in consumer_props.items():
        score = 0
        if props['max_total_value'] > 300:
            score += 2
        if props['units_p20'] >= 5:
            score += 2
        if props['elasticity'] > -1.0:
            score += 1
        if props['max_willingness'] > 35:
            score += 1
        if score >= 4:
            group_high.append(customer)
        elif score >= 2:
            group_medium.append(customer)
        else:
            group_low.append(customer)
    return group_high, group_medium, group_low
def group_demand_at_price(group, price):
    total_quantity = 0
    for customer in group:
        customer_demand = 0
        for q in range(0, 25):
            if df.loc[q, customer] >= price:
                customer_demand += 1
        total_quantity += customer_demand
    return total_quantity
def find_optimal_group_price(group):
    best_profit = 0
    best_price = 0
    best_quantity = 0
    for price in range(1, 51):
        quantity = group_demand_at_price(group, price)
        profit = (price - 4) * quantity
        if profit > best_profit:
            best_profit = profit
            best_price = price
            best_quantity = quantity
    return best_price, best_quantity, best_profit
consumer_props = analyze_consumer_properties()
group_high, group_medium, group_low = create_proper_groups(consumer_props)
print(f"\nГруппы потребителей:")
print(f"Высокая ценность: {group_high}")
print(f"Средняя ценность: {group_medium}")
print(f"Низкая ценность: {group_low}")
price_high, q_high, profit_high = find_optimal_group_price(group_high)
price_medium, q_medium, profit_medium = find_optimal_group_price(group_medium)
price_low, q_low, profit_low = find_optimal_group_price(group_low)
total_profit_3rd_degree = profit_high + profit_medium + profit_low
total_quantity_3rd_degree = q_high + q_medium + q_low
print(f"\n=== РЕЗУЛЬТАТЫ ===")
print(f"Группа высокой ценности:")
print(f"  - ОПТИМАЛЬНАЯ ЦЕНА: {price_high}")
print(f"\nГруппа средней ценности:")
print(f"  - ОПТИМАЛЬНАЯ ЦЕНА: {price_medium}")
print(f"\nГруппа низкой ценности:")
print(f"  - ОПТИМАЛЬНАЯ ЦЕНА: {price_low}")
print(f"\nИтог:")
print(f"  - ПРИБЫЛЬ: {total_profit_3rd_degree}")
```
